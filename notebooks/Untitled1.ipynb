```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sqlalchemy import create_engine
```

```python
query = "SELECT COUNT(*) AS total FROM iot_measurements;"

pd.read_sql(query, engine)
```
